```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time
```

```python
service = Service("C:/Users/asus/chromedriver-win64/chromedriver.exe")
```

```python
driver = webdriver.Chrome(service=service)
url = "https://www.naukri.com/data-analyst-jobs?k=data%20analyst"

driver.get(url)

time.sleep(3)

jobs = driver.find_elements(By.CSS_SELECTOR, "div.cust-job-tuple.layout-wrapper")

for job in jobs:
    title = job.find_element(By.CSS_SELECTOR, "a.title").get_attribute("title")
    print(title)
```
