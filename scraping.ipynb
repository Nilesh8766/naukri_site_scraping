```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
```

```python
service = Service("C:/Users/asus/chromedriver-win64/chromedriver.exe")
```

```python
# options = Options()
# options.add_argument("--headless-new")

driver = webdriver.Chrome(service=service)

url = "https://www.naukri.com/java-jobs?k=java"

driver.get(url)

time.sleep(3)

jobs = driver.find_elements(By.CSS_SELECTOR, "div.cust-job-tuple.layout-wrapper")
time.sleep(3)

for job in jobs:
    title = job.find_element(By.CSS_SELECTOR, "a.title").get_attribute("title")
    company = job.find_element(By.CSS_SELECTOR, "a.comp-name").get_attribute("title")
    experience = job.find_element(By.CSS_SELECTOR, "span.expwdth").get_attribute("title")
    location = job.find_element(By.CSS_SELECTOR, "span.locWdth").get_attribute("title")
    job_desc = job.find_element(By.CSS_SELECTOR, "span.job-desc").text
    job_posted = job.find_element(By.CSS_SELECTOR, "span.job-post-day").text
    print(title, company, experience, location, job_desc, job_posted)

    time.sleep(2)

driver.quit()
```
